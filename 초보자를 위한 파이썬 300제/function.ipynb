```python
#220
def print_max(x, y, z):
    maxvalue = 0
    if x > y:
        maxvalue = x
    else:
        maxvalue = y  
    if z > maxvalue:
        maxvalue = z
    print(maxvalue)
print_max(4, 5, 6)
```
